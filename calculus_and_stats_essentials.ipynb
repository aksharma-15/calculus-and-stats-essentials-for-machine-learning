## **What is a derivative?**

* The derivative of a function measures how much the function output changes when you make a small change in the input.

* They help us analyze change.
  * Prices change, population changes, temperature changes — derivatives measure how fast.

* They help us find maximum and minimum values.
  * To optimize anything — profit, accuracy, speed — derivatives are essential.

* They power machine learning optimization.
  * ML models learn by adjusting weights based on derivatives.

### **Simple Real-Life Story**

* You’re driving. Your speedometer doesn't tell where you are, it tells
how fast your position is changing.
  >**Derivative = speedometer of math.**

```python
# Code example with SymPy
from sympy import symbols, diff

# Define variable and function
x = symbols('x')
f = x**3 + 2*x**2 + 4

# Compute derivative
df_dx = diff(f, x)

# Print the results
print(f'Function: {f}')
print(f'Derivative: {df_dx}, (3x^2 + 4x)')
print(f'Slope at x = 2: {df_dx.subs(x, 2)}')
```

### Explanation of the above code cell:
* We declare a function f(x)
* diff() gives the derivative
* Substituting x = 2 shows how steep the function is.

#### Common confusion:
  > People often think a derivative gives a number — but it gives a formula that tells howchange behaves at any point.


### Key takeaways
* A derivative measures rate of change.
* It tells how steep a function is at any point.

## **Gradients**
* If derivatives deal with single-input functions, gradients handle functions with many inputs.
* Derivative → slope on a line.
* Gradient → direction of steepest increase on a surface

## **Why it matters?**
* Optimization in ML uses gradients to adjust weights.
* Gradients show which direction reduces loss fastest.
* They power gradient descent, the core of deep learning.

## **Real-World uses of gradients**
* Training neural networks → Gradients reduce loss efficiently.
* Designing recommendation systems → Gradients optimize weights.
* Robotics control → Gradients adjust motor paramters smoothly.

## **Example (Intuition)**
* Imagine you're blindfolded on a hill.
* The gradient tells you the direction to walk uphill fastest.
* For optimization, we walk downhill , so we take the negative gradient.

## **Problems gradients solve in Machine Learning**
* Machine learning models adjust many weights.
* We need a way to know which direction to move each weight to reduce error.

## **Before → After**
| **Without gradient**                   | **With gradient**                               |
|----------------------------------------|-------------------------------------------------|
| Model updates are random               | Updates move intentionally toward lower loss    |
| Slow or no learning                    | Model converges efficiently                     |
| Difficult for multi-variable functions | Works perfectly even with millions of variables |

```python
# Mini code example
from sympy import symbols, diff

x, y = symbols('x y')
f = x**2 +y**2 # simple surface

grad_x = diff(f, x)
grad_y =diff(f,y)

print(f'Gradient: {(grad_x, grad_y)}')
print(f'At point (2,3): {grad_x.subs({x:2, y:3})}, {grad_y.subs({x:2, y:3})}')
```

### **Explanation of the above code cell**
* Gradient = direction of steepest increase.
* To minimize → move in the negative of that direction.

### **Common confusion**
* A common confusion is thinking that gradient = slope just like derivatives.

  * *Clarification*
    > * A derivative is slope in 1 dimension, but a gradient is a vector of slopes across multiple dimensions.
    > * So, gradient doesn’t give just “how steep” — it gives “in which direction the steepest change happens.”

## **Key takeaways**
* Gradients generalize derivatives to multi-variable functions.
* Gradient direction = steepest increase.
* Gradient Descent = walk in negative gradient direction.

## **SymPy for Symbolic Math**

### What is Sympy?
* SymPy is a Python library that lets you do math like you're writing on paper — not numeric approximation, but actual algebra.

* What it does:
  * Automates derivative steps — no need to differentiate manually.
  * Ensures correctness when solving symbolic math problems.
  * Helps verify model equations and loss derivatives in ML.

### SymPy → problem it solves
Manual differentiation is slow and prone to mistakes.

**Before → After**

| **Without SymPy**      | **With SymPy**                     |
|------------------------|------------------------------------|
| Human error in algebra | Computer algebra = always correct  |
| Hard to check steps    | Steps are visible and reproducible |
| Time-consuming         | Instant symbolic results           |


* *Note*
    > Symbolic math (exact) and numerical math (approximate) are different. SymPy deals with symbolic math.

### Key takeaways
* SymPy gives exact results.
* Useful for learning derivatives and gradients transparently.
* Helps debug ML loss function math.

```python
# Example 1
from sympy import symbols, diff

x = symbols('x')
f = x**2 + 3*x + 5 # function f(x)
derivative = diff(f, x) #  derivative → f'

print(f'Derivative: {derivative}')
```

```python
# Example 2
from sympy import symbols, sin, diff

x = symbols('x')
f = sin(x)*x**2

df = diff(f, x)

print(f'Function: {f}')
print(f'Derivative: {df}')
```

## **Descriptive Statistics**

* Raw data is overwhelming. We need ways to summarize and understand it.

#### **Before → After**
| **Raw Data**        | **Summary Statistics**        |
|---------------------|-------------------------------|
| Hard to interpret   | Clear and compact description |
| Outliers hidden     | Outliers become visible       |
| No idea of patterns | Trends immediately visible    |

```python
# Example 3
import numpy as np

data = np.array([12, 15, 18, 20, 30, 35, 50])

mean = np.mean(data)
median = np.median(data)
variance = np.var(data)

print(f' Mean: {mean}, Median: {median}, Variance: {variance}')
```

#### Explanation of the above code cell
* `mean` gives the average value → a single number representing the center of the data.
* `median` gives the middle value → helps when data has outliers (like 50).
* `variance` measures how spread out the values are → higher variance = more variation.

#### In the dataset:
* The `mean` will be pulled upward because of the large number 50.
* The `median` is more stable, showing typical performance.
* The `variance` shows that the data is not tightly clustered — the values vary a lot.

#### Common confusion
Students often think mean = typical value, but when there are outliers, median is actually a more reliable measure of the “typical” value.

#### Key takeaways
* Mean ≠ Median when data is skewed.
* Variance tells how spread data is.
* First step of any data science project.

## **Probability Distributions**
#### What are probability distributions?
* They describe how likely each outcome is in a random process.
  > Examples:
    * Normal distributions
    * Uniform distributions
    * Bernouli distributions

#### Why they matter?
* They model uncertainty and risk.
* Form the foundation of statistical inference.
* Essential for probabilistic machine learning models.

#### Normal distribution:
      * Mean = 0
      * Most values cluster near the center

### Probability → Problem it solves
* We need to predict the likelihood of events in uncertain conditions.

### **Before → After**

| **Without Distributions**  | **With Distributions**           |
|----------------------------|----------------------------------|
| Randomness feels confusing | Uncertainty becomes measurable   |
| Hard to model outcomes     | Easy to compute probabilities    |
| No generalization          | Can simulate & forecast behavior |

#### Analogy:
> Weather predictions:
  * We can’t say exact rain, but we can say there is a 70% chance

```python
# Example 4
import numpy as np

# Generate random samples from normal distribution
samples = np.random.normal(loc = 50, scale = 5, size = 1000)

print(f'Mean: {np.mean(samples)}')
print(f'Std dev: {np.std(samples)}')
```

### Explanation of the above code cell
* We generated 1,000 values centered around 50 (mean).
* scale=5 means most values lie within ±5 units of the mean.
* The mean and standard deviation of the generated sample should be close to 50 and 5, because that’s how normal distributions behave.

### This demonstrates how probability distributions let us model uncertainty.
### Even though individual values differ, the overall pattern is predictable and follows the bell curve (normal or gaussian distribution).

### Common confusion:
* Many beginners think distributions are about single exact numbers.
* But distributions describe the pattern of many possible outcomes, not one event.

### Key takeaways:
* Distributions define likelihood of outcomes.
* Many natural things follow the normal distribution.
* Understanding distributions improves model interpretability.

## **Central Limit Theorem (CLT)**
* The CLT states that:
  > When you average many random variables, the average tends to follow a normal distribution, even if the original data wasn’t normal.

* Real world applications of CLT:
  * Election polls → Average opinion becomes normally distributed.
  * Manufacturing → Average defect rate stabilizes.
  * A/B testing → Average click-through behaves predictably.

### Example:
* If we survey many people'e daily screen-time:
  * Even though individuals vary wildly,
  * The average screen time across groups will form a bell curve.

### Central Limit Theorem → Problem it solves
* Individual data may be chaotic, but averages behave predictably.

### **Before → After**
| **Looking at raw data** | **Looking at averages**       |
|-------------------------|-------------------------------|
| Messy and irregular     | Forms a normal distribution   |
| Hard to compare groups  | Can compare easily & reliably |
| Unstable                | Averages stabilize            |

```python
# Example 5
import numpy as np
import matplotlib.pyplot as plt

sample_means = []
for _ in range(500):
  sample = np.random.randint(1, 100, size = 30)
  sample_means.append(np.mean(sample))

plt.figure(figsize=(6,4))
plt.hist(sample_means)
plt.show()
```

### Explanation of the above code
* Even though individual values are random, the mean forms a smooth bell curve → CLT.

### Common Confusion
* Students think data must be normal — false.
* CLT applies to the average, not raw data.

### Key takeaways
* CLT explains why we can trust averages.
* Makes statistical inference mathematically valid.
* One of the most important theorems in data science.
